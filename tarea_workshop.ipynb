#Pilares POO

Pregunta teórica:
¿Cuáles son los 4 pilares de la programación orientada a objetos y qué aporta cada uno?

```
Los 4 pilares de POO son: Abstraccion, Polimorfismo, Encapsulamiento, Herencia

*   Abstraccion: Define los atributos, y funciones (métodos) de una clase. Simplifica la complejidad del objeto
*   Polimorfismo:  Hace que una misma instruccion dada sea interpretada de forma distinta por diferentes objetos
*   Encapsulamiento: Hace que el contenido interno de la clase sea protegido.
*   Herencia: Hace que una clase pueda heredar métodos y atributos de otra clase base.   
```

````python
#Ejercicio:
#Define una jerarquía simple para vehículos con al menos una clase base y dos clases hijas.
#Cada clase hija debe tener un método propio sobrescrito que imprima información
#diferente. Crea una función que reciba un vehículo y llame a ese método.
#Agregá encapsulamiento con atributos privados y métodos get y set.
class vehiculo:
    def __init__(self, asientos, patente):
        self.asientos = asientos
        self.patente = patente

    def imprimir_info(self, auto):
        return f"{auto.marca} {auto.modelo} {auto.anio}"

    def get_asientos(self):
        return self.asientos

    def get_patente(self):
        return self.patente

    def set_asientos(self, asientos):
        self.asientos = asientos

    def set_patente(self, patente):
        self.patente = patente




class pickup(vehiculo):
    def __init__(self, marca, modelo, anio, asientos, patente, preciousd):
        super().__init__(asientos, patente)
        self.marca = marca
        self.modelo = modelo
        self.anio  = anio
        self.preciousd = preciousd

    def get_marca(self):
        return self.marca

    def get_modelo(self):
        return self.modelo

    def get_anio(self):
        return self.anio

    def get_precio(self):
        return self.preciousd

    #-------------------------------------------------------------------------------
    def set_marca(self, marca):
        self.marca = marca

    def set_modelo(self, modelo):
        self.modelo = modelo

    def set_anio(self, anio):
        self.anio = anio

    def set_precio(self, preciousd):
        self.preciousd = preciousd

    #---------------------------------------------------------------------------------

    def imprimir_info(self):
        return f"{self.marca} {self.modelo} {self.anio} con {self.asientos} asientos. Precio ${self.preciousd} USD"



class auto(vehiculo):
    def __init__(self, marca, modelo, anio, asientos, patente, preciousd):
        super().__init__(asientos, patente)
        self.marca = marca
        self.modelo = modelo
        self.anio  = anio
        self.preciousd = preciousd

    def get_marca(self):
        return self.marca

    def get_modelo(self):
        return self.modelo

    def get_anio(self):
        return self.anio

    def get_precio(self):
        return self.preciousd

    #-------------------------------------------------------------------------------
    def set_marca(self, marca):
        self.marca = marca

    def set_modelo(self, modelo):
        self.modelo = modelo

    def set_anio(self, anio):
        self.anio = anio

    def set_precio(self, preciousd):
        self.preciousd = preciousd

    #---------------------------------------------------------------------------------

    def imprimir_info(self):
        return f"El auto es un {self.marca} {self.modelo} {self.anio}. Cuenta con {self.asientos} asientos, y cuesta ${self.preciousd} USD"
````

#decoradores

Pregunta teórica:
¿Qué es un decorador en Python y para qué se utiliza comúnmente?


```
Un decorador es una funcion, a la cual se le da como argumento otra función, que modifica su comportamiento.
```

✅Mini autoevaluación
1. ¿Como se aplica a una función?
2. ¿Qué función interna suele tener un decorador?

```
1.   Un decorador se aplica llamando a la funcion original con un @.
     Ej: @MyOriginalFunct
     y debajo de esa llamada se escribe la funcion que se va a pasar como argumento.

2.   La funcion interna que suele tener un decorador es el wrapper
```

````python
#Código con errores para corregir

def decorator(func):
 print("Decorating...")
 def wrapper():
    return func()
 return wrapper()

@decorator
def greet():
 print("Hi!")
````

````python
"""Ejercicio práctico:
Crea un decorador @authorize que solo permita ejecutar una función si un parámetro user
tiene el atributo is_admin=True. Si no, debe imprimir “Acceso denegado”. Prueba el
decorador con una función de ejemplo."""

def authorize(func):
    is_admin = True
    def wrapper(function, admin):
        if admin is True:
            return function()
        else:
            print("Acceso denegado")

    return wrapper(func, is_admin)

@authorize
def Message():
    print("Hello")
````
